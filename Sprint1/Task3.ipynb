<h2 style="color:DodgerBlue">Описание проекта:</h2>----Развивается проект Sprint0 по варианту №12. Базовый класс `Item` описывает товар, а производные `SingleItem`, `PackageItem` и `SpecialItem` расширяют его свойства и поведение.В Sprint1 добавлены конструкторы с проверкой данных, свойства с геттерами и сеттерами, изменение цены, расчёт скидки, возврат товара и взаимодействие объектов через `ShoppingCart` и `Order`.#### Дополнительные функции- добавление и удаление товаров из корзины;- расчёт общей стоимости и применение скидок;- оформление заказа;- возврат товара с восстановлением его доступности.

```
using System;using System.Collections.Generic;using System.Linq;class Item{    public int ItemId { get; }    public string Name { get; }    public decimal Price { get; protected set; }    public bool IsReturned { get; private set; }    public Item(int itemId, string name, decimal price)    {        if (itemId <= 0) throw new ArgumentOutOfRangeException(nameof(itemId));        if (string.IsNullOrWhiteSpace(name)) throw new ArgumentException("Название не может быть пустым", nameof(name));        if (price < 0) throw new ArgumentOutOfRangeException(nameof(price));        ItemId = itemId; Name = name; Price = price;    }    public virtual string GetDetails() => $"Товар №{ItemId}: {Name}, цена {Price:C}";    public virtual decimal CalculateDiscount() => Price * 0.05m;    public virtual void ApplyDiscount(decimal percent)    {        if (percent < 0 || percent > 100) throw new ArgumentOutOfRangeException(nameof(percent));        Price *= 1 - percent / 100;    }    public void ReturnItem() => IsReturned = true;}class SingleItem : Item{    public string UnitMeasure { get; set; }    public SingleItem(int id, string name, decimal price, string unitMeasure) : base(id, name, price) => UnitMeasure = unitMeasure;    public override string GetDetails() => $"{base.GetDetails()}, единица: {UnitMeasure}";}class PackageItem : Item{    public int QuantityPerPackage { get; set; }    public PackageItem(int id, string name, decimal price, int quantity) : base(id, name, price) => QuantityPerPackage = quantity;    public override string GetDetails() => $"{base.GetDetails()}, в упаковке: {QuantityPerPackage} шт.";    public override decimal CalculateDiscount() => Price * 0.05m * QuantityPerPackage;}class SpecialItem : Item{    public DateTime DiscountExpirationDate { get; set; }    public SpecialItem(int id, string name, decimal price, DateTime expirationDate) : base(id, name, price) => DiscountExpirationDate = expirationDate;    public override string GetDetails() => $"{base.GetDetails()}, скидка до: {DiscountExpirationDate:dd.MM.yyyy}";    public override void ApplyDiscount(decimal percent)    {        if (DateTime.Today <= DiscountExpirationDate) base.ApplyDiscount(percent);    }}class ShoppingCart{    private readonly List<Item> items = new();    public IReadOnlyList<Item> Items => items;    public void Add(Item item) { if (!item.IsReturned) items.Add(item); }    public bool Remove(int itemId) => items.RemoveAll(item => item.ItemId == itemId) > 0;    public decimal GetTotal() => items.Where(item => !item.IsReturned).Sum(item => item.Price);    public void ApplyDiscount(decimal percent) { foreach (var item in items.Where(item => !item.IsReturned)) item.ApplyDiscount(percent); }}class Order{    public int OrderId { get; }    public string CustomerName { get; set; }    public ShoppingCart Cart { get; }    public bool IsConfirmed { get; private set; }    public Order(int orderId, string customerName, ShoppingCart cart) { OrderId = orderId; CustomerName = customerName; Cart = cart; }    public void Confirm() { IsConfirmed = true; Console.WriteLine($"Заказ №{OrderId} оформлен для {CustomerName}. Сумма: {Cart.GetTotal():C}"); }}var milk = new SingleItem(1, "Молоко", 89.90m, "литр");var tea = new PackageItem(2, "Чай", 350m, 20);var coffee = new SpecialItem(3, "Кофе", 499m, DateTime.Today.AddDays(7));var cart = new ShoppingCart();cart.Add(milk); cart.Add(tea); cart.Add(coffee);foreach (var item in cart.Items) Console.WriteLine(item.GetDetails());cart.ApplyDiscount(10);var order = new Order(1001, "Лаврентьев Владислав", cart);order.Confirm();coffee.ReturnItem();Console.WriteLine($"Товар '{coffee.Name}' возвращён: {coffee.IsReturned}");Console.WriteLine($"Сумма после возврата: {cart.GetTotal():C}");
```
